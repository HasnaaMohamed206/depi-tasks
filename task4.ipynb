Write a class `Circle` with an `__init__` that takes a `radius`, and a method `area()` that returns the circle's area (`3.14159 * radius ** 2`). Create a circle with radius 5 and print its area rounded to 2 decimals.

### Exercise 6 — Class using a dictionary internally
Write a class `ShoppingCart` with an `items` dictionary (item name → price). Add a method `add_item(item, price)` and a method `total()` that returns the sum of all prices. Add two items and print the total.

```python
class ShoppingCart:
    def __init__(self):
        self.items = {}

    def add_item(self, item, price):
        self.item = item
        self.price = price

        self.items[self.item] = self.price

    def total(self):
        total = 0
        for value in self.items.values():
            total += value

        return total
```

```python
cart = ShoppingCart()

cart.add_item("banana", 2)
cart.add_item("apple", 1.5)
cart.add_item("milk", 4)
cart.add_item("pepsi", 0.5)

cart.items
```

```python
cart.total()
```

### Exercise 12 — Implementing a Queue
Write a class `Queue` with an `items` list. Add `enqueue(item)` (adds to the end) and `dequeue()` (removes and returns the *first* item, or `None` if empty — note this is different from a Stack!). Enqueue "a", "b", "c", then dequeue once and print both the result and the remaining queue.

```python
class Queue:
    def __init__(self):
        self.items = []

    def enqueue(self, item):
        self.items.append(item)

    def dequeue(self):
        if len(self.items) == 0:
            return None
        return self.items.pop()
```

```python
queue = Queue()

queue.enqueue("a")
queue.enqueue("b")
queue.enqueue("c")

queue.items
```

```python
queue.dequeue()
queue.dequeue()

queue.items
```

### Exercise 13 — Objects containing other objects
Write a class `Movie` with `title` and `rating`. Write a class `MovieCollection` with a `movies` list, a method `add(movie)`, and a method `average_rating()` that returns the average rating of all movies (or 0 if empty). Add 3 movies with ratings 8, 6, 10 and print the average.

```python
class Movie:
    def __init__(self, title, rating):
        self.title = title
        self.rating = rating

class MovieCollection:
    def __init__(self):
        self.movies = []

    def add(self, movie):
        self.movies.append(movie)

    def average_rating(self):
        if len(self.movies) == 0:
            return 0

        total = sum(movie.rating for movie in self.movies)
        return total / len(self.movies)
```

```python
collection = MovieCollection()

collection.add(Movie("Movie 1" , 9.8))
collection.add(Movie("Movie 2" , 7.4))
```

```python
collection.average_rating()
```

### Exercise 15 — Inventory system with validation
Write a class `Inventory2` with a `stock` dictionary. `add_stock(item, qty)` should add to the existing quantity (or create it if new). `remove_stock(item, qty)` should only succeed (and return `True`) if there's enough stock, otherwise return `False` without changing anything. Add 50 then 20 more of "pen", try removing 30 (should succeed), then try removing 100 (should fail).

```python
class Inventory:
    def __init__(self):
        self.stock = {}

    def add_stock(self, item, qty):
        if item in self.stock:
            self.stock[item] += qty
        else:
            self.stock[item] = qty

    def remove_stock(self, item, qty):
        if item in self.stock and self.stock[item] >= qty:
            self.stock[item] -= qty
            return True
        return False
```

```python
good = Inventory()
good.add_stock("pen", 50)

good.stock
```

```python
good.remove_stock("pen", 30)
```

```python
good.remove_stock("pen", 100)
```

### Exercise 18 — Tracking a running high score
Write a class `Game` with `score` and `high_score`, both starting at 0. `add_points(pts)` should increase `score`, and update `high_score` if `score` beats it. `reset()` should set `score` back to 0 **without** touching `high_score`. Add 50, then 30, then reset, then add 20 — print the final `score` and `high_score`.

```python
class Game:
    def __init__(self, score = 0, high_score = 0):
        self.score = score
        self.high_score = high_score

    def add_points(self, pts):
        self.score += pts

        if self.score > self.high_score:
            self.high_score = self.score

    def reset(self):
        self.score = 0
```

```python
game = Game()
game.score
```

```python
game.add_points(50)
game.score
```

```python
game.add_points(30)
game.score
```

```python
game.reset()

print(f"The Current Score is: {game.score} , The High Score is: {game.high_score}")
```

### Exercise 20 — Comprehensive: a simple linked list
Write a class `Node2` with a `value` and a `next` (defaulting to `None`). Write a class `LinkedListSimple` with a `head` (starting as `None`) and a method `add(value)` that creates a new node and appends it to the end of the list (walking through `.next` links to find the end). Add a method `to_list()` that walks the list and returns a normal Python list of all the values, for easy printing. Add 10, 20, 30 and print the result of `to_list()`.

```python
class Node():
    def __init__(self, value, next = None):
        self.value = value
        self.next = next


class LinkedListSimple:
    def __init__(self):
        self.head = None

    def add(self, value):
        new_node = Node(value)

        if self.head is None:
            self.head = new_node
            return

        current = self.head
        while current.next is not None:
            current = current.next

        current.next = new_node

    def to_list(self):
        result = []
        current = self.head

        while current is not None:
            result.append(current.value)
            current = current.next

        return result
```

```python
linked_list = LinkedListSimple()

linked_list.add(10)
linked_list.add(20)
linked_list.add(30)

linked_list.to_list()
```
