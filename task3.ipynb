```python
import math
```

```python
def circle(r) :
    print(f"Info about a circle when r = {r} is: ")

    area_of_circle = math.pi * r * r
    perimeter_of_circle = 2 * math.pi * r

    print(f"\tThe Perimeter is: {perimeter_of_circle:.3f}")
    print(f"\tThe Area is: {area_of_circle:.3f}")
```

```python
circle(7)
```

```python
def add_all_nums(*nums) :
    for num in nums :
        if type(num) != int and type(num) != float :
            return"The Arguments must be numbers"

    return(sum(nums))
```

```python
def convert_celsius_2_fahrenheit (degree_in_selsius) :
    degree_in_fahrenheit = (degree_in_selsius * 9 / 5) + 32

    print(f"The degree is Fahrenheit is: {degree_in_fahrenheit}")
```

#### 1.4

```python
def check_season (month) :
    if month == "September" or month == "October" or month == "November" :
        print("The season is Autumn")

    elif month == "December" or month == "January" or month == "February" :
        print("The season is Winter")

    elif month == "March" or month == "April" or month == "May" :
        print("The season is Spring")

    elif month == "June" or month == "July" or month == "August" :
        print("The season is Summer")

    else:
        print("Not Valid")
```

```python
check_season("April")
```

```python
print(calculate_slope(1, 2, 3, 6))
```

#### 1.6

```python
def solve_quadratic_eqn(a, b, c):
    d = b**2 - 4*a*c

    x1 = (-b + math.sqrt(d)) / (2*a)
    x2 = (-b - math.sqrt(d)) / (2*a)

    return x1, x2
```

```python
print(solve_quadratic_eqn(1, -5, 6))
```

```python
def sum_of_numbers(num):
    sum = num

    while num > 0 :
        num -= 1
        sum += num

    print(sum)
```

```python
sum_of_numbers(5)
```

```python
sum_of_numbers(10)
```

```python
sum_of_numbers(100)
```

#### 1.12

```python
def sum_of_odd(num) :
    sum = 0

    if num % 2 == 0 :
        num  = num - 1

    while num > 0:
        sum += num
        num -= 2

    print(f"Sum of Odd Numbers less than {num} is: {sum} ")
```

```python
sum_of_odd(5)
```

```python
sum_of_odd(6)
```

```python
def sum_of_even(num) :
    sum = 0

    if num % 2 == 1 :
        num  = num - 1

    while num > 0:
        sum += num
        num -= 2

    print(f"Sum of Even Numbers less than {num} is: {sum} ")
```

```python
sum_of_even(6)
```

```python
sum_of_even(7)
```
