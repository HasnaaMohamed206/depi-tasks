```python
import math
```

```python
def circle(r) :
    print(f"Info about a circle when r = {r} is: ")

    area_of_circle = math.pi * r * r
    perimeter_of_circle = 2 * math.pi * r

    print(f"\tThe Perimeter is: {perimeter_of_circle:.3f}")
    print(f"\tThe Area is: {area_of_circle:.3f}")
```

```python
circle(7)
```

```python
def check_season (month) :
    if month == "September" or month == "October" or month == "November" :
        print("The season is Autumn")

    elif month == "December" or month == "January" or month == "February" :
        print("The season is Winter")

    elif month == "March" or month == "April" or month == "May" :
        print("The season is Spring")

    elif month == "June" or month == "July" or month == "August" :
        print("The season is Summer")

    else:
        print("Not Valid")
```

```python
check_season("April")
```

```python
def solve_quadratic_eqn(a, b, c):
    d = b**2 - 4*a*c

    x1 = (-b + math.sqrt(d)) / (2*a)
    x2 = (-b - math.sqrt(d)) / (2*a)

    return x1, x2
```

```python
print(solve_quadratic_eqn(1, -5, 6))
```

```python
def sum_of_odd(num) :
    sum = 0

    if num % 2 == 0 :
        num  = num - 1

    while num > 0:
        sum += num
        num -= 2

    print(f"Sum of Odd Numbers less than {num} is: {sum} ")
```

```python
sum_of_odd(5)
```

```python
sum_of_odd(6)
```

```python
def reverse_list(List):
    num_items = len(List)
    reverse_list = []

    while num_items > 0 :
        reverse_list.insert(0 , List[len(List) - num_items])

        num_items -= 1

    return reverse_list
```

```python
print(reverse_list([1, 2, 3, 4, 5]))
```

```python
print(reverse_list(['A', 'B', 'C']))
```

```python
def evens_and_odds(num) :
    if num < 0 :
        return "The number must be positive"

    sum_odd = 0
    sum_even = 0

    while num >=0 :
        if num % 2 == 0 :
            sum_even += 1
        else :
            sum_odd +=1

        num -=1

    print(f"The number of odd are: {sum_odd}")
    print(f"The number of even are: {sum_even}")
```

```python
evens_and_odds(100)
```

```python
def calculate_mean(List) :
    sum = 0

    for item in List :
        sum += item

    print(f"The Mean is: {sum / len(List)}")
```

```python
calculate_mean([1,2,3,4,5])
```

```python
def calculate_median(List) :
    List.sort()
    index =(len(List) // 2)

    if len(List) % 2 == 1 :
        return List[index]

    else :
        return(List[index - 1] + List[index]) / 2
```

```python
print(calculate_median([5, 2, 4, 1, 3]))
```

```python
print(calculate_median([5, 2, 4, 1, 3, 6]))
```

#### 3.1

```python
def is_prime(num) :
    if num < 2 : 
        return False

    for i in range(2, num) :
        if num % i == 0 :
            return False

    return True
```

```python
print(is_prime(5))
```

```python
print(is_prime(9))
```

```python
print(is_prime(1))
```
