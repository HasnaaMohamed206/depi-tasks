## Part One

```python
age = int(input("Enter your Age: "))

if age >= 18 :
    print("You are old enough to drive")

else :
    print(f"You need {18 - age} more years to learn to drive")
```

```python
age = int(input("Enter your Age: "))

my_age = 21

if age == my_age + 1:
    print(f"You are 1 year older than me")

elif my_age < age :
    print(f"You are {age - my_age} years older than me")

elif my_age == age + 1:
    print(f"I am 1 year older than you")

elif my_age > age :
    print(f"I am {my_age - age} years older than you")

else :
    print("We are in the same age")
```

```python
curr_month = input("Enter the current month: ")

if curr_month == "September" or curr_month == "October" or curr_month == "November" :
    print("The season is Autumn")

elif curr_month == "December" or curr_month == "January" or curr_month == "February" :
    print("The season is Winter")

elif curr_month == "March" or curr_month == "April" or curr_month == "May" :
    print("The season is Spring")

elif curr_month == "June" or curr_month == "July" or curr_month == "August" :
    print("The season is Summer")

else:
    print("Not Valid")
```
